```python
import lxml
import requests
from urllib.parse import urljoin

import requests
from requests.packages.urllib3.util.retry import Retry
from requests.adapters import HTTPAdapter

from io import StringIO

from lxml.html import parse
from itertools import islice



base_url = "https://opencontext.org"


```

```python
def root_doc_for_content(content):
    doc = parse(StringIO(content)).getroot()
    return doc

def root_doc_for_url(url):
    content = content_for_url(url)
    return root_doc_for_content(content)
```

```python
class Scraper(object):
    def __init__(self, base_url="https://opencontext.org"):
        self.base_url = base_url
        self.session = requests.Session()
        retries = Retry(total=5,
                backoff_factor=0.3,
                status_forcelist=[ 500, 502, 503, 504 ])
        self.session.mount(self.base_url, HTTPAdapter(max_retries=retries))
        headers = {
            'User-Agent': 'https://raymondyee.net',
            'From': 'raymond.yee@gmail.com'  # This is another valid field
        }
        self.session.headers.update(headers)
    def projects(self):
        url = urljoin(base_url, "/projects-search/")
        
        while url:
            r = self.session.get(url)
            doc = root_doc_for_content(r.text)
            rows = doc.xpath('//*[@id="panel-geo-recs"]/div/table/tbody/tr')
            yield from rows
            try:
                # look for url for forward button
                url = doc.xpath('//span[contains(@class,"glyphicon-forward")]/parent::a/@href')[0]
            except:
                url = None
```

```python
s = Scraper('http://octest.raymondyee.net')
prjs = list(s.projects())
```
